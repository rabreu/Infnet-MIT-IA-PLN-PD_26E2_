```python
from sklearn.feature_extraction.text import CountVectorizer # TF
from sklearn.feature_extraction.text import TfidfVectorizer # TF-iDF
from sklearn.metrics.pairwise import cosine_similarity
```

```python
corpus = [
    "O estudo de bias algorítmico é importante.",
    "Algoritmos podem conter bias ocultos.",
    "A inteligência artificial aprende com os dados.",
    "Dados ruins geram modelos com bias e erros.",
    "O estudo da ética na inteligência artificial é fundamental."
]

print(f"Corpus carregado com { len(corpus) } documentos")
#
```

# Stopwords

```python
from nltk.stem.snowball import stopwords
# !pip install -q spacy nltk
import nltk
nltk.download("stopwords")

stopwords_nltk = stopwords.words("portuguese")
```

# Bag of Words

```python
vectorizer = CountVectorizer(max_df=0.8,
                             lowercase=True,
                             stop_words=stopwords_nltk)

X_bow = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X_bow.toarray())
```

```python
import pandas as pd

pd.DataFrame(X_bow.toarray(), columns=vectorizer.get_feature_names_out())
```

# TF-iDF

```python
vectorizer = TfidfVectorizer(max_df=0.8,
                             min_df=1,
                             lowercase=True,
                             stop_words=stopwords_nltk)

X_tfidf = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X_tfidf.toarray())
```

```python
pd.DataFrame(X_tfidf.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
query = ["Como identificar bias em inteligência artificial?"]
```

```python
# 1. Transformar a query usando o MESMO vectorizer que a gente usou antes

query_vec = vectorizer.transform(query)

pd.DataFrame(query_vec.toarray(), columns=vectorizer.get_feature_names_out())


# 2. Calcular a similaridade de cosse entre a query e todos os documentos

similaridades = cosine_similarity(query_vec, X_tfidf).flatten()

# 3. Rankear os resultados
resultados = pd.DataFrame({"documento": corpus, "similaridade": similaridades})


# 4. Ordenar os resultados pelos resultados mais próximos
resultados.sort_values(by="similaridade", ascending=False, inplace=True)

resultados
```
