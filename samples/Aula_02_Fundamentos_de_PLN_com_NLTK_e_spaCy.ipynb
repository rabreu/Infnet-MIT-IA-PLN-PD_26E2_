```python
import spacy

# !python -m spacy download en_core_web_md
# !python -m spacy download pt_core_news_md
```

```python
nlp_en = spacy.load("en_core_web_md")
nlp_pt = spacy.load("pt_core_news_md")
```

```python
text_en = """
Trump warned that Iranian forces would be
 "blown off the face of the Earth" if they target US ships.
Meanwhile, Iran's attacks on the UAE threaten to
widen the conflict again.
"""

text_pt = """
O Flamengo oficializou nesta segunda-feira a renovação de
contrato do atacante Bruno Henrique.
O novo vínculo é válido até o fim de 2027
e foi assinado após o treino no Ninho do Urubu.
O vínculo anterior se encerrava em dezembro deste ano.
"""

doc_pt = nlp_pt(text_pt)
doc_en = nlp_en(text_en)


for idx, token in enumerate(doc_pt):
    print(f"Token {idx}: {token}")
```

```python
def mostrar_tokens_spacy(doc):
    dados = []
    for token in doc:
        dados.append({
            "token": token,
            "é pontuação": token.is_punct,
            "é espaço": token.is_space,
            "é alfa numérico": token.is_alpha,
            "formato": token.shape_
        })
    return dados

import pandas as pd

pd.DataFrame(mostrar_tokens_spacy(doc_pt))
```

# Limpar o texto

```python
from nltk.corpus import stopwords

```

```python
from nltk.corpus import stopwords

def limpar_texto(text, language="english", custom_stopwords=[]):

    stopwords_ = set(stopwords.words(language) + custom_stopwords)
    tokens = [token.text.lower() for token in nlp_pt(text)]
    tokens = [
        token
        for token in tokens
        if token not in stopwords_ and token.isalpha()
    ]
    return tokens

limpar_texto(text_pt, language="portuguese")
```

```python
text_pt = """
Programação em python é muito melhor que em javascript.
Quem programa em python sabe o que é programação divertida.
"""
tokens = limpar_texto(text_pt, language="portuguese", custom_stopwords=["sabe"])

for tuple_tokens in Counter(tokens).most_common(10):
    print(tuple_tokens)
```

# Stemming com Porter e Snowball

Reduz as palavras flexonadas para radical

```python
from nltk.stem import PorterStemmer, SnowballStemmer

stemmer_porter = PorterStemmer()
stemmer_snowball = SnowballStemmer(language="portuguese")
```

```python
porter = PorterStemmer()

palavras_en = ["running", "runs", "runner", "studies", "studying", "houses", "better"]

for palavra in palavras_en:
    print(palavra, "->", porter.stem(palavra))
```

```python
snowball_pt = SnowballStemmer("portuguese")

palavras_pt = [
    "correr",
    "correndo",
    "correu",
    "corrida",
    "casas",
    "casinha",
    "felizmente",
    "programação",
    "programador",
    "programando"
]

for palavra in palavras_pt:
    print(palavra, "->", snowball_pt.stem(palavra))
```

```python
text_pt = """
Programação em python é muito melhor que em javascript.
Quem programa em python sabe o que é programação divertida.
"""
tokens = limpar_texto(text_pt, language="portuguese", custom_stopwords=["sabe"])
tokens_stem = [stemmer_snowball.stem(token) for token in tokens]
for tuple_tokens in Counter(tokens_stem).most_common(10):
    print(tuple_tokens)
```

# Lematização

Palavras flexonadas -> lema (representação no dicionário)

```python
text_pt = """
Programação em python é muito melhor que em javascript.
Quem programa em python sabe o que é programação divertida.
"""

doc = nlp_pt(text_pt)

stopwords_ = set(stopwords.words("portuguese") )

tokens = [
    token.lemma_.lower()
    for token in doc
]

tokens = [
    token
    for token in tokens
    if token not in stopwords_ and token.isalpha()
]

for tuple_tokens in Counter(tokens).most_common(10):
    print(tuple_tokens)
```

```python
frase = "Os jogadores estavam correndo e comemorando a vitória no estádio."

doc = nlp_pt(frase)

comparacao = []

for token in doc:
    if token.is_alpha:
        palavra = token.text.lower()
        stem = snowball_pt.stem(palavra)
        lema = token.lemma_

        comparacao.append({
            "palavra": palavra,
            "stemming": stem,
            "lematização": lema,
            "classe_gramatical": token.pos_
        })

pd.DataFrame(comparacao)
```
