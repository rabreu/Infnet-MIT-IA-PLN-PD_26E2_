```python
# !pip install gensim nltk pyLDAVis pandas

import pandas as pd

import nltk
from nltk.corpus import stopwords
from gensim.utils import simple_preprocess
from sklearn.datasets import fetch_20newsgroups

nltk.download('stopwords')

stop_words = stopwords.words('english')
stop_words.extend(
    ['from', 'subject', 're', 'edu', 'use', 'organization', 'cc', 'writes']
)


categories = ['sci.space','comp.graphics', 'talk.religion.misc', 'sci.med']
newsgroups = fetch_20newsgroups(subset='all', categories=categories, remove=('headers', 'footers', 'quotes'))
```

```python
data = pd.DataFrame({'text': newsgroups.data, 'target': newsgroups.target})
data
```

```python
def clean_text(text):
    tokens = simple_preprocess(str(text), deacc=True, min_len=3)
    return [token for token in tokens if token not in stop_words]

data['tokens'] = data['text'].apply(clean_text)

data = data[data['tokens'].map(len) > 0]
```

```python
data
```

```python
import gensim.corpora as corpora
from gensim.models.ldamodel import LdaModel

id_dict = corpora.Dictionary(data['tokens'])


corpus = [id_dict.doc2bow(text) for text in data['tokens']]


lda_model = LdaModel(
    corpus=corpus,
    id2word=id_dict,
    num_topics=4,
    chunksize=100,
    passes=20,
    per_word_topics=True,
    random_state=42
)

for idx, topic in lda_model.print_topics(-1):
    print('Topic: {} \nWords: {}'.format(idx, topic))
```
